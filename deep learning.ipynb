# **Neural Network (Dense) Using TensorFlow / Keras**

A **Dense Layer** (also called a **Fully Connected Layer**) is one of the most commonly used layers in neural networks.

![Dense](https://dasarpai.com/assets/images/dspost/cv/dense-layer-neural-layer.png)

In a dense layer:

* Every input neuron is connected to every output neuron
* Each connection has a learnable value called a **weight**
* During training, the network adjusts these weights to improve predictions


## **What Is a Neural Network?**

A **Neural Network** is a machine learning model that learns patterns from data.

Instead of following manually written rules, the model learns automatically by looking at examples.

The network:

* Takes input data
* Processes it through layers
* Produces predictions as output

As training continues, the model improves its predictions by reducing errors.

## **Simple Idea Behind Neural Networks**

A neural network learns in three simple steps:


### **1. Take Input Data**

The model receives data as input.

Examples:

* Age
* Job Profile
* Location


### **2. Make Predictions**

The network performs mathematical calculations on the input data to generate an output or prediction.

Example:

* Predicting Salary

### **3. Learn From Errors**

The model compares its prediction with the actual correct answer.

If the prediction is wrong:

* The network adjusts its internal values
* Tries to make better predictions next time

This process repeats many times until the model learns the pattern in the data.


## **Main Components of a Neural Network**

| Component               | Simple Explanation                               |
| ----------------------- | ------------------------------------------------ |
| **Neuron**              | Small unit that performs calculations            |
| **Weights**             | Values that decide how important each input is   |
| **Bias**                | Extra value added to improve learning            |
| **Activation Function** | Helps the network learn complex patterns         |
| **Layers**              | Groups of neurons connected together             |
| **Loss Function**       | Measures how wrong the prediction is             |
| **Optimizer**           | Updates weights to reduce errors during training |


## **The Math Inside One Dense Layer**


Each Keras `Dense` layer performs a mathematical operation on the input data.

The layer calculates:

$$\mathbf{y} = \mathbf{x} \cdot W^{T} + \mathbf{b}$$

Where:

| Symbol | Meaning                             |
| ------ | ----------------------------------- |
| `x`    | Input data given to the layer       |
| `W`    | Weights learned during training     |
| `b`    | Bias values learned during training |
| `y`    | Output produced by the layer        |


### **Understanding the Working**

A Dense layer:

* Takes input values
* Multiplies them with weights
* Adds bias values
* Produces an output

During training:

* The network keeps adjusting the weights and biases
* This helps improve predictions over time


### **Using Dense Layers in Keras**

Keras provides Dense layers using:

`tf.keras.layers.Dense(units, activation=...)`

Where:

* `units` → number of neurons in the layer
* `activation` → activation function used by the layer


### **Example Neural Network Structure**

```text
Input Features
       ↓
Dense Layer (32 neurons, ReLU)
       ↓
Hidden Layer 1
       ↓
Dense Layer (16 neurons, ReLU)
       ↓
Hidden Layer 2
       ↓
Dense Layer (1 neuron)
       ↓
Output Prediction
```


### **Why Multiple Layers?**

Using multiple dense layers helps the network learn:

* Simple patterns in early layers
* More complex patterns in deeper layers

Activation functions like `ReLU` add non-linearity, allowing the model to learn complex relationships from data.


## **Why Activation Functions Are Important**

If we stack only linear layers, the entire neural network behaves like a single linear equation.

This means:

* Multiple layers would act like one simple layer
* The network would not learn complex patterns properly

Activation functions solve this problem by adding **non-linearity** to the network.

Because of this, the model can learn:

* Curves
* Complex relationships
* Non-linear patterns in data


### **ReLU (Rectified Linear Unit)**

One of the most commonly used activation functions is **ReLU**.

It is defined as:

$$\text{ReLU}(x)=\max(0,x)$$



### **How ReLU Works**

| Input Value    | Output                    |
| -------------- | ------------------------- |
| Negative value | `0`                       |
| Positive value | Same value passes through |

Examples:

| Input | Output |
| ----- | ------ |
| `-5`  | `0`    |
| `2`   | `2`    |
| `10`  | `10`   |


### **Why ReLU Is Commonly Used**

ReLU is popular because it:

* Is simple and fast
* Helps neural networks train efficiently
* Works well for most deep learning tasks

ReLU is commonly used in hidden layers.


### **Output Layer Activation**

For regression problems, the output layer usually has **no activation function**.

Reason:

* The model should be able to predict any numerical value
* ReLU would block negative outputs

In this notebook:

* Hidden layers use `activation='relu'`
* Output layer uses no activation function


## **Loss Functions: Measuring Prediction Error**

After the network makes predictions, it needs a way to measure how correct or incorrect those predictions are.

This is the job of the **Loss Function**.

A loss function compares:

* Predicted values
* Actual target values

It then returns a single number called the **loss**.

Smaller loss means:

* Better predictions
* Better model performance

Training aims to reduce this loss value as much as possible.

### **Mean Squared Error (MSE)**

One of the most commonly used loss functions for regression problems is **Mean Squared Error (MSE)**.

It is defined as:

$$\mathcal{L}=\frac{1}{N}\sum_{i=1}^{N}(\hat{y}_i-y_i)^2$$

### **Understanding the Formula**

| Symbol | Meaning                 |
| ------ | ----------------------- |
| `ŷ`   | Predicted value         |
| `y`    | Actual value            |
| `N`    | Total number of samples |
| `L`    | Final loss value        |

The formula:

1. Finds the difference between prediction and actual value
2. Squares the difference
3. Averages all errors



### **Loss Functions for Other Problems**

Different tasks use different loss functions.

Examples:

| Problem Type               | Common Loss Function       |
| -------------------------- | -------------------------- |
| Regression                 | `Mean Squared Error (MSE)` |
| Binary Classification      | `BinaryCrossentropy`       |
| Multi-class Classification | `CategoricalCrossentropy`  |


## **The Optimizer**

After calculating the loss, the neural network needs to improve its predictions.

This is done using an **Optimizer**.

An optimizer updates the model’s weights to reduce prediction error.


### **Gradient Descent Idea**

The model checks:

* Which weights caused errors
* How much each weight should change

Weights are updated using:

$$w_{\text{new}} = w_{\text{old}} - \eta \cdot \frac{\partial \mathcal{L}}{\partial w}$$


### **Understanding the Formula**

| Symbol  | Meaning                                 |
| ------- | --------------------------------------- |
| `w`     | Weight value                            |
| `L`     | Loss                                    |
| `∂L/∂w` | Gradient (direction of error reduction) |
| `η`     | Learning rate                           |

The optimizer changes the weights step-by-step to reduce the loss.


### **What Is Adam?**

**Adam** is one of the most commonly used optimizers in deep learning.

It improves standard gradient descent by:

* Adjusting learning speed automatically
* Using information from previous updates
* Helping the model train faster and more smoothly

Because of this, Adam works well for many neural network problems.


### **Learning Rate**

The **learning rate** controls how large each weight update should be.

| Learning Rate | Effect                       |
| ------------- | ---------------------------- |
| Too large     | Training may become unstable |
| Too small     | Training becomes very slow   |

Choosing a proper learning rate is important for good training.


## **Problem Setup: Learning From Data Points**

Imagine many points plotted on a graph.

Each point contains:

* An input value `x`
* An output value `y`

The neural network only sees these points.

Its goal is to learn the hidden relationship between the inputs and outputs.


This type of problem is called **Regression**.

### **What Is Regression?**

Regression means:

> Predicting continuous numerical values from data.

Common examples include:

* Predicting house prices
* Predicting temperatures
* Predicting sales revenue

In this notebook:

* The dataset is created using `sin(x)` with added noise
* The network does not know the original formula
* It must learn the pattern only from the `(x, y)` data points

### **How the Neural Network Learns**

The learning process happens step-by-step during training.


#### **Step 1: Start With Random Weights**

At the beginning:

* The network has random weight values
* Predictions are inaccurate

The model has not learned anything yet.

#### **Step 2: Make Predictions**

The network takes input values `x` and predicts output values `y`.

These predictions are usually poor initially.

#### **Step 3: Measure the Error**

The predicted values are compared with the actual target values.

The difference between them is called the **loss** or **error**.

Smaller loss means better predictions.

#### **Step 4: Update the Weights**

The optimizer adjusts the weights slightly to reduce the error.

The network slowly improves after every update.

#### **Step 5: Repeat the Process**

The same process repeats many times during training.

Over time:

* Loss decreases
* Predictions improve
* The network learns the pattern in the data

This entire learning process is called **Training** or **Fitting**.


### **What Does the Network Learn?**

The network learns the overall shape or trend in the data.

It does not memorize every point individually.

Instead, it learns:

* How inputs and outputs are related
* How to predict values for new unseen inputs

## **Generating Synthetic Training Data**

The formula used to generate dataset is:
$$y = \sin(x) + \varepsilon, \quad \varepsilon \sim \mathcal{N}(0,0.1)$$


### **Understanding the Formula**

| Symbol   | Meaning                        |
| -------- | ------------------------------ |
| `sin(x)` | True underlying pattern        |
| `ε`      | Random noise added to the data |
| `y`      | Final target value             |

The sine wave represents the actual pattern we want the network to learn.

The added noise makes the data look more realistic and less perfect.


### **What the Network Sees**

The neural network does not know the actual formula.

It only sees:

* Input values `X`
* Output values `y`

Its job is to learn the hidden relationship between them.

### **Why This Is a Good Learning Example**

A sine wave is a non-linear pattern.

A simple linear model cannot fit this curve properly.

This demonstrates why neural networks need:

* Hidden layers
* Non-linear activation functions like `ReLU`

to learn complex patterns.


### **Data Generation Steps**

#### **Step 1: Generate Input Values**

`np.linspace(-π, π, 200)`

Creates:

* 200 evenly spaced `x` values
* Between `-π` and `π`


#### **Step 2: Generate Target Values**

`sin(x) + noise`

Creates:

* Smooth sine-wave targets
* With added random noise


#### **Step 3: Reshape the Data**

The data is reshaped to:

`(200, 1)`

Reason:

* Keras Dense layers expect input shape as:

`[samples, features]`


#### **Step 4: Convert to float32**

The data is converted to `float32`.

Reason:

* Standard datatype used in TensorFlow
* Faster and more memory efficient for neural network training


### **Why Use Synthetic Data?**

Instead of loading a real-world dataset, we create our own data.

This helps us focus on:

* How the neural network works
* How training happens
* How the model learns patterns


````python
N_SAMPLES = 200

# 200 evenly spaced x values across one full sine cycle
X_np = np.linspace(-np.pi, np.pi, N_SAMPLES).astype(np.float32)

# y = sin(x) + small Gaussian noise  →  simulates real-world imperfection
noise = (0.1 * np.random.randn(N_SAMPLES)).astype(np.float32)
y_np  = np.sin(X_np) + noise

# Reshape to [samples, features] — required by Keras Dense layers
X = X_np.reshape(-1, 1)   # shape (200, 1)
y = y_np.reshape(-1, 1)   # shape (200, 1)

print(f"X shape : {X.shape}   dtype: {X.dtype}")
print(f"y shape : {y.shape}   dtype: {y.dtype}")

# Visualise
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(X_np, np.sin(X_np), color='#4c8eda', linewidth=2, label='True sin(x)')
ax.scatter(X_np, y_np, color='#e07b4f', s=15, alpha=0.7, label='Noisy samples')
ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('y', fontsize=12)
ax.set_title('Training Data sin(x) with Gaussian Noise', fontsize=14, fontweight='bold')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3, linestyle='--')
ax.set_facecolor('#f8f9fa')
fig.patch.set_facecolor('#ffffff')
plt.tight_layout()
plt.show()
````

## **Defining the Neural Network with Keras**


Keras provides `tf.keras.Sequential` to build neural networks layer-by-layer.

In a Sequential model:

* Layers are added one after another
* Output from one layer becomes input to the next layer

This makes model creation simple and beginner-friendly.


### **Understanding a Dense Layer**

Each `Dense` layer is defined using:

| Parameter     | Purpose                                                  |
| ------------- | -------------------------------------------------------- |
| `units`       | Number of neurons in the layer                           |
| `activation`  | Activation function used by the layer                    |
| `input_shape` | Shape of the input data (only needed in the first layer) |


### **Network Architecture**

```text id="d0y9zy"
Input (1)
    ↓
Dense Layer (32 neurons, ReLU)
    ↓
Dense Layer (16 neurons, ReLU)
    ↓
Dense Layer (1 neuron)
    ↓
Output Prediction
```


#### **Why Is the Input Size `1`?**

Each training sample contains:

* One input value `x`

So the input layer needs:

* One input feature


#### **Why Is the Output Size `1`?**

The network predicts:

* One continuous numerical value `y`

So the output layer contains:

* One neuron


#### **Why Use Hidden Layers?**

A single Dense layer can only learn simple linear relationships.

Our dataset follows a sine-wave pattern, which is non-linear.

Hidden layers help the network learn:

* Curves
* Complex patterns
* Non-linear relationships


#### **Why Use ReLU in Hidden Layers?**

ReLU introduces non-linearity into the model.

Without activation functions:

* Multiple Dense layers would behave like one simple linear layer

ReLU helps the network learn more complex patterns effectively.


#### **Why No Activation in the Output Layer?**

The output layer predicts continuous numerical values.

Using ReLU would remove negative outputs.

Since sine values can be both:

* Positive
* Negative

the output layer uses:

* No activation function


#### **Why Use `32 → 16` Neurons?**

The network gradually reduces layer size:

`32 → 16 → 1`

This structure helps:

* Learn important features in earlier layers
* Compress information toward the final prediction

This type of structure is commonly used in neural networks.

### **Understanding `model.summary()`**

Keras provides:

`model.summary()`

This displays:

* All layers in the model
* Output shape of each layer
* Number of trainable parameters

The total parameter count includes:

* Weights
* Bias values

from all Dense layers in the network.


````python
from tensorflow import keras
from tensorflow.keras import layers

# Define the network
model = keras.Sequential([
    layers.Dense(32, activation='relu', input_shape=(1,)),  # Layer 1: 1 → 32
    layers.Dense(16, activation='relu'),                    # Layer 2: 32 → 16
    layers.Dense(1),                                        # Output:  16 → 1  (no activation)
], name='sine_regressor')

# Print full architecture + parameter counts
model.summary()
````

## **Compiling the Model**


Before training starts, the model needs to be configured using:

`model.compile()`

Compilation tells Keras:

* How the model should learn
* How errors should be measured
* What information should be displayed during training

### **Compilation Settings**

| Argument    | Value     | Purpose                                     |
| ----------- | --------- | ------------------------------------------- |
| `optimizer` | `'adam'`  | Updates weights during training             |
| `loss`      | `'mse'`   | Measures prediction error for regression    |
| `metrics`   | `['mae']` | Displays additional performance information |

### **Optimizer: Adam**

`'adam'` is one of the most commonly used optimizers.

It:

* Updates weights efficiently
* Helps the model converge faster
* Automatically adjusts learning behavior during training

### **Loss Function: MSE**

`'mse'` stands for **Mean Squared Error**.

It is commonly used for:

* Regression problems
* Continuous numerical predictions

MSE measures how far predictions are from actual target values.

### **Metrics: MAE**

`'mae'` stands for **Mean Absolute Error**.

It calculates:

* The average absolute difference between predictions and actual values

MAE is easier to interpret because it shows the error directly in the same units as the target values.

### **Important Note**

`model.compile()` does **not** start training.

It only prepares the model by attaching:

* Optimizer
* Loss function
* Evaluation metrics

````python
 
model.compile(
    optimizer='adam',     # Adam: adaptive learning rates + momentum
    loss='mse',           # Mean Squared Error, standard for regression
    metrics=['mae'],      # Mean Absolute Error shown in training logs
)

print("Optimizer :", model.optimizer.get_config()['name'])
print("Loss      :", model.loss)
````

## **Training the Model**

`model.fit()` is used to train the neural network.

It automatically handles the complete training process internally.

During training, the model repeatedly:

* Makes predictions
* Calculates error
* Updates weights
* Improves performance over time


### **Training Loop**

Internally, each training step follows this process:

```text id="1v6qoq"
Clear Old Gradients
        ↓
Forward Pass
        ↓
Calculate Loss
        ↓
Backward Pass
        ↓
Update Weights
```


### **Important Arguments in `model.fit()`**

| Argument     | Value         | Purpose                                |
| ------------ | ------------- | -------------------------------------- |
| `X, y`       | Training data | Input features and target values       |
| `epochs`     | `2000`        | Number of complete training cycles     |
| `batch_size` | `200`         | Number of samples processed together   |
| `verbose`    | `0`           | Hides training output during execution |


### **Understanding Each Argument**

#### **Training Data: `X, y`**

* `X` contains input values
* `y` contains target output values

The model learns the relationship between them.


#### **Epochs**

`epochs=2000`

An epoch means:

> One complete pass through the entire dataset.

More epochs allow the model to:

* Learn patterns gradually
* Improve predictions over time


#### **Batch Size**

`batch_size=200`

This means:

* All 200 samples are processed together in one step

This is called:

* **Full Batch Gradient Descent**


#### **Verbose**

`verbose=0`

This suppresses training logs during execution.

Instead of printing every epoch:

* Results will be visualized later using plots.


### **History Object**

`model.fit()` returns a **History object**.

It stores training information such as:

* Loss values
* MAE values

for every epoch during training.

This history is usually stored as:

`history`

and later used for plotting training curves.


### **What Happens During Each Epoch?**

#### **1. Forward Pass**

Input data `X` moves through all Dense layers.

The network produces predictions.


#### **2. Loss Calculation**

The model compares:

* Predicted values
* Actual target values `y`

The difference becomes the loss.


#### **3. Backward Pass**

TensorFlow automatically computes gradients for all weights.

These gradients show:

* How weights should change
* To reduce prediction error


#### **4. Weight Update**

The Adam optimizer updates the weights using:

* Gradients
* Previous update information

This helps the model learn efficiently.

#### **Training Outcome**

As training continues:

* Loss decreases
* Predictions improve
* The network gradually learns the sine-wave pattern from the data


````python
# Train
EPOCHS = 2000

history = model.fit(
    X, y,
    epochs=EPOCHS,
    batch_size=200,  
    verbose=0,        
)

loss_history = history.history['loss']

print(f"Initial loss : {loss_history[0]:.4f}")
print(f"Final loss   : {loss_history[-1]:.4f}")
print(f"Reduction    : {(1 - loss_history[-1] / loss_history[0]) * 100:.1f}%")
````

## **Visualising the Training Loss Curve**

After training a neural network, one of the first things to check is the **loss curve**.

The loss curve shows:

* How the loss changed during training
* Whether the model learned properly
* How training progressed over epochs

In this notebook, the curve represents:

* Mean Squared Error (MSE)
* Recorded after every epoch

````python
# Loss Curve
fig, ax = plt.subplots(figsize=(10, 4))

ax.plot(range(1, EPOCHS + 1), loss_history, linewidth=1.8,
        color='#4c8eda', label='Training Loss (MSE)')

ax.set_xlabel('Epoch', fontsize=13)
ax.set_ylabel('MSE Loss', fontsize=13)
ax.set_title('Training Loss over Epochs', fontsize=15, fontweight='bold', pad=14)
ax.legend(fontsize=12)
ax.grid(True, alpha=0.3, linestyle='--')
ax.set_facecolor('#f8f9fa')
fig.patch.set_facecolor('#ffffff')

plt.tight_layout()
plt.savefig('loss_curve.png', dpi=150, bbox_inches='tight')
plt.show()
print("Loss curve saved to loss_curve.png")
````

## **Inference and Weight Inspection**

 `model.predict()`

Keras handles inference mode automatically inside `model.predict()` — no need to call `.eval()` or `torch.no_grad()` manually.  
It disables any layers that behave differently during training (e.g. Dropout, BatchNorm) and skips gradient computation.

### **Weight Inspection**

Every `Dense` layer stores two learned tensors accessible after training:
- `layer.get_weights()[0]` — the weight matrix `W`
- `layer.get_weights()[1]` — the bias vector `b`

Inspecting these is a quick sanity check that training actually modified the weights from their random initialisation.


````python
# ── Inference ────────────────────────────────────────────────────────────
test_points = np.array([[-np.pi], [-np.pi/2], [0.0], [np.pi/2], [np.pi]], dtype=np.float32)
predictions = model.predict(test_points, verbose=0)
true_values = [0.0, -1.0, 0.0, 1.0, 0.0]

print("Inference results")
print("-" * 45)
print(f"{'Input x':<18} {'Predicted':>10} {'True sin(x)':>12}")
print("-" * 45)
for x_val, pred, true in zip(test_points.flatten(), predictions.flatten(), true_values):
    print(f"{round(float(x_val), 4):<18} {pred:>10.4f} {true:>12.4f}")

# Weight Inspection
print()
first_layer = model.layers[0]
W, b = first_layer.get_weights()
print(f"Layer 0  name   : {first_layer.name}")
print(f"Layer 0  W shape: {W.shape}")
print(f"Layer 0  b shape: {b.shape}")
print()
print("First 5 bias values after training:")
print(b[:5])
````

## Plotting the Fitted Curve

Final check: overlay the network's predictions on the original data.  
A well-trained network should trace through the middle of the noisy dots, recovering the underlying `sin(x)` shape.


````python
# Fitted Curve
X_plot = np.linspace(-np.pi, np.pi, 300).astype(np.float32).reshape(-1, 1)
y_pred = model.predict(X_plot, verbose=0)

fig, ax = plt.subplots(figsize=(10, 4))
ax.scatter(X_np, y_np, color='#e07b4f', s=15, alpha=0.5, label='Noisy training data')
ax.plot(X_np, np.sin(X_np), color='#aaaaaa', linewidth=1.5, linestyle='--', label='True sin(x)')
ax.plot(X_plot, y_pred, color='#4c8eda', linewidth=2.5, label='Network prediction')
ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('y', fontsize=12)
ax.set_title('Learned Curve vs True sin(x)', fontsize=14, fontweight='bold')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3, linestyle='--')
ax.set_facecolor('#f8f9fa')
fig.patch.set_facecolor('#ffffff')
plt.tight_layout()
plt.show()
````
